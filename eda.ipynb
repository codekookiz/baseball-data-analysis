```python
import pandas as pd
```

```python
df = pd.read_csv("open/data/train.csv")
df.head()
```

```python
df.select_dtypes(include='object').columns
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
```

```python
le1 = LabelEncoder()
df['top_bottom'] = le1.fit_transform(df['top_bottom'])

le2 = LabelEncoder()
df['game_type'] = le2.fit_transform(df['game_type'])

le3 = LabelEncoder()
df['base_state'] = le3.fit_transform(df['base_state'])
```

```python
group_mean = df.groupby('pitcher_id')['asof_pitcher_prev3_game_middle_rate'].transform('mean')

df['asof_pitcher_prev3_game_middle_rate'] = df['asof_pitcher_prev3_game_middle_rate'].fillna(group_mean)
```

```python
df.loc[:, df.isna().any()].isna().sum()
```

```python
pitcher_cols = [
    'asof_pitcher_success_rate',
    'asof_pitcher_reverse_rate',
    'asof_pitcher_middle_rate',
    'asof_pitcher_ball_rate',
    'asof_pitcher_strike_rate',
    'asof_pitcher_prev1_game_success_rate',
    'asof_pitcher_prev3_game_success_rate',
    'asof_pitcher_prev5_game_success_rate',
    'asof_pitcher_prev1_game_middle_rate',
    'asof_pitcher_prev3_game_middle_rate',
    'asof_pitcher_prev5_game_middle_rate',
    'asof_pitcher_fastball_rate',
    'asof_pitcher_breaking_rate',
    'asof_pitcher_offspeed_rate'
]

for col in pitcher_cols:
    pitcher_mean = df.groupby('pitcher_id')[col].transform('mean')
    
    global_mean = df[col].mean()
    
    df[col] = df[col].fillna(pitcher_mean).fillna(global_mean)
```

```python
batter_cols = [
    'asof_batter_success_rate',
    'asof_batter_middle_rate'
]

for col in batter_cols:
    batter_mean = df.groupby('batter_id')[col].transform('mean')
    
    global_mean = df[col].mean()
    
    df[col] = df[col].fillna(batter_mean).fillna(global_mean)
```

```python
df.iloc[:, 1:]
```

```python
sc = StandardScaler()
scaled_df = df.iloc[:, 1:]
transform_df = pd.DataFrame(sc.fit_transform(scaled_df), columns=scaled_df.columns)
df.loc[:, scaled_df.columns] = transform_df
```

```python
import numpy as np

corr = df.iloc[:, 1:].corr().abs()

upper_mask = np.triu(np.ones(corr.shape), k=1).astype(bool)
upper_corr = corr.where(upper_mask)

high_corr_pairs = (
    upper_corr.stack()
    .reset_index()
    .rename(columns={'level_0': 'Feature_1', 'level_1': 'Feature_2', 0: 'Correlation'})
)

high_corr_pairs = high_corr_pairs[high_corr_pairs['Correlation'] >= 0.5]
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.max_rows', 50)

sorted_high_corr = high_corr_pairs.sort_values(by='Correlation', ascending=False)

print(sorted_high_corr.head(20))
```
